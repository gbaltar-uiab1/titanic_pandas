###  ``` df.iloc[] ```

Selects rows and columns using integers. Supports 
index, index array and slicing 
- ``` df.iloc[0] ``` Shows the first row
- ``` df.iloc[[0, 1, 4]] ``` Shows the 1st, 2nd and 5th rows
- ``` df.iloc[:3] ``` Shows the first 3 rows (indices 0, 1, 2)

````python

# Example: Selecting columns

# Single column selection
print(df["Age"].head())

# Multiple column selection
print(df[["Sex", "Age", "Survived"]].head())
````

````python
# 
# Cf. https://pandas.pydata.org/pandas-docs/stable/reference/api/pandas.DataFrame.loc.html

# We're going to establish the percentages of Survivors and Deceased passengers.
#  This is simple aggregated information as provided by the Dataset. We'll drill down further later on.

'''
We could've used a simple row selection, boolean comparison and '.sum' to add up the resulting rows, thus;
    total_survivors = (df["Survived"] == 1).sum()
    total_deceased = (df["Survived"] == 0).sum()
However, for the sake of this exercise we'll be using '.loc' and '.shape'. 
```df.loc[ <boolean mask> ]``` selects matching rows. ``` .shape[0] ``` yields the row count.
'''


total_pax = len(df)
total_survivors = df.loc[df["Survived"] == 1].shape[0]
total_deceased  = df.loc[df["Survived"] == 0].shape[0]

print("Total passengers aboard the Titanic: ", total_pax)
print(f"Total survivors: {total_survivors}. ({(total_survivors / total_pax * 100):.2f}% of total passengers)")
print(f"Total deceased: {total_deceased}. ({(total_deceased / total_pax * 100):.2f}% of total passengers)")
````

````python
'''
- ``` df.iloc[0] ``` Shows the first row
- ``` df.iloc[[0, 1, 4]] ``` Shows the 1st, 2nd and 5th rows
- ``` df.iloc[:3] ``` Shows the first 3 rows (indices 0, 1, 2)
'''

# Since 'Sex' and 'Age' columns are next to each other, we will make a small matrix sex_age_groups
#  first we slice both columns, then we fill up age_groups
#  we'll also check whether there are any data points left that do not satisfy the conditions (ie. (Male OR Female) AND (Child OR Adult) )


sex_age_groups = df.iloc[:, [1, 4, 5]]
print(sex_age_groups)


# how to simplify this?? Matrix??
# male_adult_survivor = sex_age_groups[sex_age_groups["Sex" == "male" ] and sex_age_groups["Age" >= 16] and sex_age_groups["Survived" == 1]].count


male     = sex_age_groups["Sex"] == "male"
female   = sex_age_groups["Sex"] == "female"
adult    = sex_age_groups["Age"] >= 16
child    = sex_age_groups["Age"] < 16
survived = sex_age_groups["Survived"] == 1
deceased = sex_age_groups["Survived"] == 0


male_adult_survivor      = sex_age_groups[male & adult & survived].shape[0]
female_adult_survivor    = sex_age_groups[female & adult & survived].shape[0]

male_adult_deceased      = sex_age_groups[male & adult & deceased].shape[0]
female_adult_deceased    = sex_age_groups[female & adult & deceased].shape[0]

male_child_survivor      = sex_age_groups[male & child & survived].shape[0]
female_child_survivor    = sex_age_groups[female & child & survived].shape[0]

male_child_deceased      = sex_age_groups[male & child & deceased].shape[0]
female_child_deceased    = sex_age_groups[female & child & deceased].shape[0]


print(f"Male adult survivor tally:\t{male_adult_survivor}")
print(f"Female adult survivor tally:\t{female_adult_survivor}")

print(f"Male adult deceased tally:\t{male_adult_deceased}")
print(f"Female adult decesed tally:\t{female_adult_deceased}")

print(f"Male child survivor tally:\t{male_child_survivor}")
print(f"Female child survivor tally:\t{female_child_survivor}")

print(f"Male child deceased tally:\t{male_child_deceased}")
print(f"Female child deceased tally:\t{female_child_deceased}")


# Plot deceased and survivor number for each category using a stacked bar chart.

import matplotlib.pyplot as plt

# Transform the variables into a dictionary of lists (we need this to create a Dataframe)

segmented_data = {
    'Group' : ['Male Adult', 'Female Adult', 'Male Child', 'Female Child'],

    # Survivor counts (aligned with 'Group')
    'Survivor' : [male_adult_survivor, female_adult_survivor, male_child_survivor, female_child_survivor],

    # Deceased counts (aligned with 'Group')
    'Deceased' : [male_adult_deceased, female_adult_deceased, male_child_deceased, female_child_deceased]
}


# Create the Dataframe
df_segmented_data = pd.DataFrame(segmented_data)

df_segmented_data.head()

# Use 'Group' as Index
df_segmented_data = df_segmented_data.set_index('Group')


# Initialise Matplotlib plot
fig, ax = plt.subplots(figsize=(8, 6))

# Plot the base of the stack (Deceased count values)
p1 = ax.bar(df_segmented_data.index, df_segmented_data['Deceased'], label='Deceased', color='#d62728')

# Plot the top of the stack (Survivor count values)
p2 = ax.bar(df_segmented_data.index, df_segmented_data['Survivor'], bottom=df_segmented_data['Deceased'], label='Survivor', color='#2ca02c')

# Write out labels and chart title
ax.set_ylabel('Count')
ax.set_title('Titanic Fate by Sex and Age Group (Stacked Bar Chart)')
ax.legend(title='Fate')


# Tidy up chart layout
plt.xticks(rotation=0)
plt.tight_layout()

# Save the figure
plt.savefig('titanic_stacked_bar_chart.png')





# female_adult_survivor
# male_adult_deceased
# female_adult_deceased
# male_child_survivor
# female_child_survivor
# male_child_deceased
# female_child_deceased

# Check if Total Count tallies to Total Pax (otherwise there's sth wrong in data!)
````

From the above Stacked Chart we can easily infer that Male Adult casualties took the bulk of casualties. 
This is a supporting argument to the investigation's claim that the evacuation policy of *Women and children first* was largely observed.

However, we haven't done any basic checking yet! Let's compare the number of total passengers within the dataframe 'df_segmented_data' and compare it with the total passengers (earlier we saved the variable 'total_pax').

````python
# We'll calculate total_segmented_pax from the lists in the dataframe 'df_segmented_data'
# Then we'll compare this value with 'total_pax' to ascertain whether the previous chart is reliable.

print("QUALITY CONTROL:")
print("=================\n")

total_segmented_pax = df_segmented_data['Survivor'].sum() + df_segmented_data['Deceased'].sum()
print(f"Total Segmented Pax: \t{total_segmented_pax} \t---> (all included in above Chart)")
print(f"Total Dataset Pax: \t{total_pax} \t---> (Total Dataset headcount)\n")

if total_segmented_pax < total_pax:
    print("Total Segmented Pax count is LOWER than Total Dataset Pax count!\nAnalysis is not 100 per cent reliable!")
    print(f"  Ratio of segmented pax vs total pax: {(total_segmented_pax / total_pax * 100):.2f}%")

elif total_segmented_pax > total_pax:
    print("Total Segmented Pax count is HIGHER than Total Dataset Pax count!\nAnalysis is not 100 per cent reliable!")
    print(f"  Ratio of segmented pax vs total pax: {(total_segmented_pax / total_pax * 100):.2f}%")

else:
    print("All registered passengers are included in the above Bar Chart")
    
    
````

Using the ```.query()``` method let's check the survival rate of Females in 1st class.
- Pclass is the strongest survival predictor in the Titanic's sinking.
- 1st class passengers have the highest survival rate due to having cabins closest to deck.
- The evacuation protocol gave priority to women and children.

We shall break down all females traveling 1st class as either **Unaccompanied** (traveling alone) or **Accompanied** (wives, daughters, sisters or mothers traveling with relatives), regardless of age. This splitting is meant as a simple way to have a test group —Conveniently for us, *Unaccompanied* amounts for about 1 third of the 1st Class female subset, while *Accompanied* contains the remaining two thirds.

Thus, if the above premises are true, both groups  should have a similar ratio of survivors and deceased.
To put this to the test, we chose a Mortality Rate (or Death Rate) metric.

--> The results are indeed very close; **2.94%** and **3.33%** for the *Unaccompanied* and *Accompanied* groups, respectively.

````python


# Let's assign vars for Unaccompanied female surviving and deceased pax in 1st class

unaccompanied_Fem_1st_survivors = df.query("Sex == 'female' and Survived == 1 and Pclass == 1 and SibSp == 0 and Parch == 0")
unaccompanied_Fem_1st_deceased = df.query("Sex == 'female' and Survived == 0 and Pclass == 1 and SibSp == 0 and Parch == 0")


# Get counts:
unaccompanied_Fem_1st_survivors_count = unaccompanied_Fem_1st_survivors.shape[0]
unaccompanied_Fem_1st_deceased_count = unaccompanied_Fem_1st_deceased.shape[0]


# Let's assign vars for Accompanied female surviving and deceased pax in 1st class

accompanied_Fem_1st_survivors = df.query("Sex == 'female' and Survived == 1 and Pclass == 1 and (SibSp > 0 or Parch > 0)")
accompanied_Fem_1st_deceased = df.query("Sex == 'female' and Survived == 0 and Pclass == 1 and (SibSp > 0 or Parch > 0)")


# Get counts:
accompanied_Fem_1st_survivors_count = accompanied_Fem_1st_survivors.shape[0]
accompanied_Fem_1st_deceased_count = accompanied_Fem_1st_deceased.shape[0]




# Mortality Rate
print("Mortality Rate (Females traveling 1st Class): [1]\n===================================================\n")
print(f"{'\033[1m'}Unaccompanied 1st class females{'\033[0m'}. Survivors: {unaccompanied_Fem_1st_survivors_count}, Deceased: {unaccompanied_Fem_1st_deceased_count}.\n\t Unaccompanied Mortality Rate:\t{(unaccompanied_Fem_1st_deceased_count / (unaccompanied_Fem_1st_deceased_count + unaccompanied_Fem_1st_survivors_count) * 100):.2f}%")
print(f"{'\033[1m'}Accompanied 1st class females{'\033[0m'}. Survivors: {accompanied_Fem_1st_survivors_count}, Deceased: {accompanied_Fem_1st_deceased_count}.\n\t Accompanied Mortality Rate:\t{(accompanied_Fem_1st_deceased_count / (accompanied_Fem_1st_deceased_count + accompanied_Fem_1st_survivors_count) * 100):.2f}%")

print(f"\n\n[1] Mortality Rate (or Death Rate) is the probability of death within each group;\n\t\t\tMortality Rate = Deceased # / Total in Category * 100")


# Odds of dying - Discarded metric
# print(f"{'\033[1m'}Unaccompanied 1st class females{'\033[0m'}. Survivors: {unaccompanied_Fem_1st_survivors_count}, Deceased: {unaccompanied_Fem_1st_deceased_count}. Deceased Ratio:\t{(unaccompanied_Fem_1st_deceased_count / unaccompanied_Fem_1st_survivors_count * 100):.2f}%")
# print(f"{'\033[1m'}Accompanied 1st class females{'\033[0m'}. Survivors: {accompanied_Fem_1st_survivors_count}, Deceased: {accompanied_Fem_1st_deceased_count}. Deceased Ratio:\t{(accompanied_Fem_1st_deceased_count / accompanied_Fem_1st_survivors_count * 100):.2f}%")
````

### Missing rows

As we mentioned earlier, the total rows in the original dataset is **891**, but the segmented tally is **714**.

The reason for this discrepancy is easily detected if we trace back the steps taken and we look at the output of the Pandas features explored so far.

There is a row that has only 714 non-null rows, i.e. 'Age'.
This information became apparent when we issued ```df.info()```.

As we Later split and filtered our data by values of 'Age',  only the 714 non-null rows were carried forward;

```adult    = sex_age_groups["Age"] >= 16```<br>
```child    = sex_age_groups["Age"] < 16```

````python
print("Showing non-null rows per column using df.info()\n")
print(df.info())
````

Let's now take action on those missing records.

```df.isna()``` shows ```True``` for each null value in the Dataframe.

```df.isna().sum()``` shows the total of Null vales for each column.

````python
# Example: Check missing values

print(df.isna())

print(df.isna().sum())
````

````python

# Fill missing values in 'Age' with column median value.

df["Age"] = df["Age"].fillna(df["Age"].median())
df["Age"].head()

# Note that 'Age' has now 891 non-null values as we've filled its null values with the column's median value.
# Note also this doesn't affect other columns with Null values, such as 'Cabin'
print(df.info())
````


```df.dropna()```

The column 'Cabin' has only 204 values. This is roughly 23% of the dataset row count.
With such small amount of valid rows we are unable to fill in the remaining 73% and make this row's information meaningful.

So, let's drop the column 'Cabin'.



````python
# try with wrong name
# df.drop('column_name', axis=1, inplace=False)

# Let's drop the column 'Cabin'
#   but only if the column still exists.

if 'Cabin' in df.columns:
    print("Column 'Cabin' exists. Deleting...")
    df.dropna('Cabin', axis=1, inplace=True)
    print("Column 'Cabin' deleted!")
else:
    print("Column 'Cabin' doesn't exist. No action taken.")

# Show resulting Dataframe's head.
df.head()
    
````
